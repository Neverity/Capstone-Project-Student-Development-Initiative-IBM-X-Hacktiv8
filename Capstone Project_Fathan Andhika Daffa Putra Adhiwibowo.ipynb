## Implementasi contact form

```python
%%writefile index.html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Personal Portfolio</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>

    <section id="hero">
        <h1>Turning Data into Decisions.</h1>
        <p>I am a passionate data scientist with a strong background in machine learning and statistical analysis.</p>
    </section>

    <section id="about">
        <h1>About Me</h1>
        <p>I am a passionate data scientist with a strong background in machine learning and statistical analysis. I enjoy transforming complex data into actionable insights and building innovative solutions. As a data enthusiast, I thrive on exploring datasets, uncovering patterns, and telling data-driven stories. My expertise lies in Python, R, SQL, and various visualization tools. With a knack for problem-solving and a love for data, I am dedicated to leveraging technology to make informed decisions and create impactful results. I'm always eager to learn and take on new challenges.</p>
    </section>

    <section id="skills">
        <h1>Skills</h1>
        <p>List your key skills here.</p>
    </section>

    <section id="projects">
        <h1>Projects</h1>
        <p>Showcase your projects here.</p>
    </section>

    <section id="contact">
        <h1>Contact Form</h1>
        <p>Include a contact form or link to your contact method here.</p>

        <!-- Basic HTML Form - To integrate with a no-backend service like Google Forms:
             1. Create a form using the service (e.g., Google Forms).
             2. Get the form's action URL and method (usually POST).
             3. Replace the 'action="#"' and 'method="post"' attributes below with the ones provided by the service.
             4. Ensure the 'name' attributes of the input fields match the field names in your service's form.
        -->
        <form action="#" method="post">
            <div>
                <label for="name">Name:</label>
                <input type="text" id="name" name="name" required>
            </div>
            <div>
                <label for="email">Email:</label>
                <input type="email" id="email" name="email" required>
            </div>
            <div>
                <label for="message">Message:</label>
                <textarea id="message" name="message" rows="4" required></textarea>
            </div>
            <div>
                <button type="submit">Send Message</button>
            </div>
        </form>

    </section>

</body>
</html>
```
